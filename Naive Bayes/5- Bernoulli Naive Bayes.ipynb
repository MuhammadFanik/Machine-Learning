```python
from sklearn.naive_bayes import BernoulliNB
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.pipeline import Pipeline

# Same emails but now we care about
# PRESENCE (1) or ABSENCE (0) of a word — not count

emails = [
    "win free money now click here",
    "congratulations you won a prize call now",
    "free offer limited time buy now",
    "claim your free reward today",
    "meeting tomorrow at 3pm please confirm",
    "project update attached please review",
    "lunch tomorrow sounds good",
    "please find the report attached",
]
labels = [1, 1, 1, 1, 0, 0, 0, 0]

pipeline_bernoulli = Pipeline([
    ('vectorizer', CountVectorizer(binary=True)),
    # ↑ binary=True converts counts to 0 or 1
    #   "free free money" → {'free':1, 'money':1}  (not 2)

    ('classifier', BernoulliNB(alpha=1.0))
])

pipeline_bernoulli.fit(emails, labels)

new_email = ["free prize click now"]
pred  = pipeline_bernoulli.predict(new_email)
proba = pipeline_bernoulli.predict_proba(new_email)

print(f"Prediction  : {'SPAM' if pred[0]==1 else 'HAM'}")
print(f"P(Ham)      : {proba[0][0]:.3f}")
print(f"P(Spam)     : {proba[0][1]:.3f}")
```
