```python
import numpy as np
from sklearn.preprocessing import StandardScaler
import pandas as pd
```

```python
df = pd.read_csv("Social_Network_Ads.csv", usecols=["Age", "EstimatedSalary", "Purchased"])
df.sample(5)
```

```python
df.shape
```

```python
X = df[["Age", "EstimatedSalary"]]
y = df["Purchased"]
```

```python
# Train Test Split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.7, random_state=0)
print(X_train.shape, X_test.shape)
```

```python
# Standardization
scaler = StandardScaler()

# Fit the scaler to the training data, it will learn the parameters
scaler.fit(X_train)

# Scale the training and testing sets
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# Check the mean
scaler.mean_
```

```python
# Converting back the numpy arrays into dataframes
X_train_scaled = pd.DataFrame(X_train_scaled, columns=X_train.columns)
X_test_scaled = pd.DataFrame(X_test_scaled, columns=X_test.columns)
```

```python
# Check the standard Deviation
scaler.scale_
```
