```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("covid_toy.csv")
df.head()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(df[["age", "gender", "fever", "cough", "city"]], df["has_covid"], train_size=0.8, random_state=0)
```

```python
simple_imputer = SimpleImputer()    # handles missing values - Fills in missing data in your dataset
ohe = OneHotEncoder()       # Converts categories into binary columns
oe = OrdinalEncoder()       # Converts categories into numbers
```

```python
# Creating a single transformer that will apply different preprocessing steps to different columns all at once.
# SimpleImputer --> Fills any missing value with the mean of the column
# OrdinalEncoder --> Converts Mild --> 0, Strong --> 1
# OHE --> Returns a regular numpy array instead of a sparse matrix
transformer = ColumnTransformer(transformers=[
    ("tnf1", SimpleImputer(), ["fever"]),
    ("tnf2", OrdinalEncoder(categories=[["Mild", "Strong"]]), ["cough"]),
    ("tnf3", OneHotEncoder(sparse_output=False, drop="first"), ["gender", "city"])
], remainder="passthrough")
```

```python
# Applying all those transformations I defined and converting the result into a pandas dataframe
# fit() --> The transformer learns the necessary stats from the data. SimpleImputer calculates the mean of the fever column.
#   OrdinalEncoder learns the mapping Mild --> 0, Strong --> 1
#   OHE learns all the unique categories in the gender and city columns

# transform() --> Applies the learned transformation to your data
# fit_transform() --> Does both step at once
# Result --> A numpy array with all transformations applied
transformed = transformer.fit_transform(df)
transformed
```

```python
# Converts the numpy array back into the pandas dataframe
transformed = pd.DataFrame(transformed)
transformed
```

```python
# Fitting and transforming only on the training data
# .fit() --> Learning phase
# .transform() --> Application Phase
X_train_transformed = transformer.fit_transform(X_train)
X_train_transformed
```

```python
# Transforming the test data using parameters learned from the training data (using stats and mapping learned from X_train)
# SimpleImputer on fever --> Uses the mean calculated from X_train
# OrdinalEncoder on cough --> Uses the mapping learned from X_train
# OHE on gender and city --> Creates the exact same columns as it did for X_train
X_test_transformed = transformer.transform(X_test)
X_test_transformed
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer

# Sample dataset
df = pd.DataFrame({
    'age': [25, 32, 47],
    'salary': [50000, 60000, 120000],
    'city': ['New York', 'Paris', 'Tokyo']
})

# Define transformations
ct = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), ['age', 'salary']),
        ('cat', OneHotEncoder(), ['city'])
    ],
    remainder='drop')

# Fit and transform
X_transformed = ct.fit_transform(df)

print("Transformed Array:\n", X_transformed)
print("Feature Names:\n", ct.get_feature_names_out())
```
