# Working with Dates

```python
# The first thing is to convert the date/time columns into date/time datatype
order.head()
```

```python
order["date"] = pd.to_datetime(order["date"])
```

```python
order.info()
```

```python
# Extract Year
order["year"] = order["date"].dt.year
order.head()
```

```python
# Extract Month
order["month"] = order["date"].dt.month
order.head()
```

```python
# Extract month name
order["month_name"] = order["date"].dt.month_name()
order.head()
```

```python
# Extract Days (which day of the month?)
order["day_of_month"] = order["date"].dt.day
order.head()
```

```python
# Day of week - (Which day of week?)
order["day_of_week"] = order["date"].dt.dayofweek
order.head()
```

```python
# Day of the week (name)
order["name_day_of_week"] = order["date"].dt.day_name()
order.head()
```

```python
# Extract Time elapsed between two dates
import datetime

today = datetime.datetime.today()
today
```

```python
today - order["date"]
```
