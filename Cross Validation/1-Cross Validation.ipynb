```python
import numpy as np
import pandas as pd
from sklearn.model_selection import cross_val_score, KFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
```

```python
# Simulate a loan default dataset
np.random.seed(42)

n_samples = 1000

data = pd.DataFrame({
   "income": np.random.normal(50000, 15000, n_samples), # Annual Income
   "age": np.random.randint(21, 65, n_samples), # Age
   "loan_amount": np.random.normal(20000, 7000, n_samples), # Loan amount
   "credit_score": np.random.randint(300, 850, n_samples),  # Credit Score
   "year_employed": np.random.randint(0, 40, n_samples),  # Work history
   "defaulted": np.random.choice([0, 1], n_samples, p=[0.7, 0.3])
})
data.head()
```

```python
# Features and target
X = data.iloc[:, 0:-1]
y = data.iloc[:, -1]
```

```python
# Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Apply cross-validation
model = RandomForestClassifier(random_state=42)

# Split the dataset into 5 folds, randomly shuffle the dataset
kf = KFold(n_splits = 5, shuffle=True, random_state=42)

# cross_val_score --> Runs cross validation
# model --> RandomForestClassifier
# X_scaled --> Feature data
# y--> target labels
# cv --> tells it to use KFold object we created
# scoring --> evaluation metric (can be f1, roc_auc, etc...)
scores = cross_val_score(model, X_scaled, y, cv=kf, scoring="accuracy")

print("Scores from each fold", scores)
print("Average accuracy score", scores.mean())
```
