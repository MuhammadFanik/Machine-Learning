```python
import numpy as np
from sklearn.datasets import load_wine  # Wine Dataset
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier # DT model
from sklearn.metrics import accuracy_score  # Model Performance
from sklearn import tree  # For visualizing the tree
import matplotlib.pyplot as plt
```

```python
# Load the Dataset
wine = load_wine()
X = wine.data # Input data
y = wine.target # Target Labels (0, 1, 2)
```

```python
# Shape
print(X.shape)
```

```python
# Split the data into train test split
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.8, random_state=42)
```

```python
# Model Building
clf = DecisionTreeClassifier()

# Fit the data
clf.fit(X_train, y_train)
```

```python
# Make Predictions on the test data
y_prediction = clf.predict(X_test)
```

```python
# Evaluate the model
accuracy = accuracy_score(y_test, y_prediction)
print(accuracy)
```

```python
# Visualize the tree
plt.figure(figsize=(12, 8))
tree.plot_tree(clf, feature_names=wine.feature_names, class_names=wine.target_names, filled=True)
plt.show()
```
