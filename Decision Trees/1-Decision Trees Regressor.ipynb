```python
from sklearn.datasets import load_diabetes  # Diabetes Dataset
from sklearn.model_selection import train_test_split  # Split the data
from sklearn.tree import DecisionTreeRegressor   # DT model
from sklearn.metrics import mean_squared_error, r2_score  # model performance
from sklearn import tree   # Visualizing the tree
import numpy as np
```

```python
# Load the dataset
diabetes = load_diabetes()
X = diabetes.data
y = diabetes.target
```

```python
print(X.shape)
```

```python
# Split the data
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.8, random_state=42)
```

```python
# Create the Decision Tree Regressor Model
dt_reg = DecisionTreeRegressor()

# Fit the model on the training data
dt_reg.fit(X_train, y_train)
```

```python
# Make Predictions
y_pred = dt_reg.predict(X_test)
```

```python
# Evaluate the model
mse = mean_squared_error(y_test, y_pred)
print(mse)
```

```python
# R2 score
r2 = r2_score(y_test, y_pred)
print(r2)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(12, 8))
tree.plot_tree(dt_reg, feature_names=diabetes.feature_names, filled=True)
plt.show()
```
