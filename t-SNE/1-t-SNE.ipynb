```python
# Importing the required libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.manifold import TSNE
```

```python
# Create synthetic dataset: 3 clusters in 3D
# Returns 2 things: X, the actual data points (a numpy array of shape 300, 3)
# the cluster labels for each point(a numpy array of shape 300,) containing labels 0, 1, 2
X, y = make_blobs(n_samples=300, n_features=3, centers=3, random_state=0)
```

```python
X
```

```python
y
```

```python
# Apply t-SNE to reduce the dimensions (3D --> 2D)
# This creates a t-SNE transformer object with specific configuration parameters
# n_components --> The number of dimensions in the output (lower-dimensional space) --> Reducing from 3D to 2D
# perplexity --> Controls how many neighbors each point considers when computing similarities
# random_state --> Sets the random seed for reproducible results. Same seed = same initial positions = same final visualization
tsne = TSNE(n_components=2, perplexity=30, random_state=0, learning_rate=200)
tsne
```

```python
# This executes the entire t-SNE algorithm on your data

# fit_transform() --> Takes your input data X(300, 3). Computes high dimensional similarities using Gaussian distributions.
# Initialized random 2D positions for all 300 points. Returns the optimized 2D coordinates
X_tsne = tsne.fit_transform(X)
```

```python
# Visualize the original 3D data (before applying t-SNE)
fig = plt.figure(figsize=(10, 4))
ax = fig.add_subplot(121, projection="3d")
ax.scatter(X[:,0], X[:,1], X[:,2], c=y, cmap="tab10")
ax.set_title("Original 3D data")
plt.show()
```

```python
# Visualize the reduced 2D Data
ax2 = fig.add_subplot(122)
ax2.scatter(X_tsne[:,0], X_tsne[:, 1], c=y, cmap="tab10")
ax2.set_title("2D representation via t-SNE")
plt.tight_layout()
plt.show()
```

```python
plt.scatter(X_tsne[:,0], X_tsne[:,1], c=y, cmap="tab10")
plt.title("Data Transformed into 2D using t-SNE")
plt.show()
```
