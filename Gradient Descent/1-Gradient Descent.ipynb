```python
# Importing the required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Generate some synthetic data
np.random.seed(42)
X = np.linspace(0, 10, 100)
y = 3 * X + 4 + np.random.randn(100)  # Add some noise
```

```python
X.shape
```

```python
y.shape
```

```python
# Initialize model parameters
m = 2
b = 5
learning_rate = 0.01
epochs = 100
n = len(X)
```

```python
# Track MSE during training
cost_history = []
```

```python
# Gradient Descent Loop
for epoch in range(epochs):
  y_predicted = m*X + b # Computes the predicted value of y based on current m and b
  error = y_predicted - y # This is the residual

  # Compute Gradients
  dm = (1/n) * np.dot(X, error)    # gradient w.r.t. slope
  db = (1/n) * np.sum(error)       # gradient w.r.t. intercept

  # Update Parameters
  m = m - (learning_rate * dm)
  b = b - (learning_rate * db)

  # Compute Cost and store it
  cost = np.mean(error**2)
  cost_history.append(cost)
```

```python
# Final Parameters after training
print(f"Final Model: y = {m:0.2f}x + {b:0.2f}")
```

```python
# Plotting the results
fig, axs = plt.subplots(1, 2, figsize=(12, 5))

# Data and the best fit line
axs[0].scatter(x=X, y=y, color="#065084", label="Data")
axs[0].plot(X, m*X+b, color="#EB5B00", label="Fitted Line")
axs[0].set_title(f"Linear Regression via Fitted Line\n y = {m:0.2f}x + {b:0.2f}")
axs[0].set_xlabel("X")
axs[0].set_ylabel("Y")
axs[0].legend()
plt.grid()
# Cost over training epochs
axs[1].plot(range(epochs), cost_history, color="#004030")
axs[1].set_title("MSE")
plt.grid()
plt.show()
```
