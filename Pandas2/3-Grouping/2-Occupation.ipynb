```python
# Step 1: Import the libraries
import pandas as pd
from scipy.datasets import ascent
```

```python
# Step 2 and 3
users = pd.read_csv("https://raw.githubusercontent.com/justmarkham/DAT8/master/data/u.user", sep="|")
users.head()
```

```python
users.shape
```

```python
# Step 4: Discover what is the mean age per occupation
round(users.groupby("occupation")["age"].mean())
```

```python
users.head()
```

```python
# Step 5: Discover the Male ratio per occupation and sort it from the most to the least
grouped = users.groupby("occupation")

male_count = grouped["gender"].apply(lambda x: (x=="M").sum())
total_count = grouped["gender"].count()
male_ratio = (male_count/total_count)*100
male_ratio.sort_values(ascending=False)
```

```python
# Step 6: For each occupation, calculate the minimum and maximum ages
users.groupby("occupation")["age"].agg(["min", "max"])
```

```python
# Step 7: For each combination of occupation and gender, calculate the mean age
round(users.groupby(["occupation", "gender"])["age"].agg(["mean"]), 1)
```
