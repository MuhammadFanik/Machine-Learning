```python
# Step 1 - Import the necessary libraries
import numpy as np
import pandas as pd
```

```python
# Step 2 and 3: Import the dataset and assign it to a variable called chipo
chipo = pd.read_csv("https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv", sep="\t")
```

```python
# Step 4: See the first 10 entries
chipo.head(10)
```

```python
# Step 5: What is the number of observations in the dataset?
chipo.shape[0]
```

```python
# Step 6: What is the number of columns in the dataset?
chipo.shape[1]
```

```python
# Step 7: Print the name of all the columns.
print(chipo.columns)
```

```python
# Step 8: How is the dataset indexed?
chipo.index
```

```python
# Step 9: Which was the most-ordered item?
c = chipo.groupby("item_name")["quantity"].sum().sort_values(ascending=False)
c.head(1)
```

```python
# Step 10: For the most-ordered item, how many items were ordered?
c = chipo.groupby("item_name")["quantity"].sum().sort_values(ascending=False)
c.head(1)
```

```python
# Step 11: What was the most ordered item in the choice_description column?
chipo["choice_description"].value_counts().sort_values(ascending=False).head(1)
```

```python
# Step 12: How many items were ordered in total?
print(chipo["quantity"].sum())
```

```python
# Step 13: Turn the item price into a float

# lambda is an anonymous function that takes in a value (e.g "$2.50") and removes the first character, then converts it into a floating number. SO the purpose is to remove the dollar sign
dollarizer = lambda x: float(x[1:])

# .apply applies the dollarizer function to every element in the item_price columns
chipo["item_price"] = chipo["item_price"].apply(dollarizer)
chipo["item_price"]
```

```python
chipo.info()
```

```python
# Step 14: How much was the revenue for the period in the dataset?
chipo.head()
```

```python
revenue = (chipo["quantity"] * chipo["item_price"]).sum()
print(f"Revenue was: ${revenue}")
```

```python
# Step 15: How many orders were made in the period?
x = chipo["order_id"].value_counts().count()
x
```

```python
# Step 16: What is the average revenue amount per order?
average_revenue_amount = revenue / x
print(f"Average Revenue amount is: ${average_revenue_amount:.2f}")
```
