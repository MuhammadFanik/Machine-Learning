```python
import numpy as np
import pandas as pd
from sklearn.feature_selection import RFE
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
# 1. Create a meaningful custom dataset (Housing data)
np.random.seed(42)
num_samples = 500

data = {
    "house_size": np.random.normal(2000, 500, num_samples),             # sqft
    "num_rooms": np.random.randint(3, 8, num_samples),                  # rooms
    "crime_rate": np.random.uniform(0, 1, num_samples),                 # 0 = safe, 1 = dangerous
    "distance_to_city_center": np.random.uniform(1, 30, num_samples),   # miles
    "school_rating": np.random.uniform(1, 10, num_samples),             # 1-10
    "age_of_house": np.random.randint(1, 80, num_samples),              # years
}

df = pd.DataFrame(data)
df.head()
```

```python
# Create a target variable: House Price
df["price"] = (
    df["house_size"] * 150
    + df["school_rating"] * 4000
    - df["crime_rate"] * 30000
    - df["age_of_house"] * 500
    - df["distance_to_city_center"] * 1500
    + np.random.normal(0, 20000, num_samples)   # noise
)
```

```python
X = df.drop("price", axis=1)
y = df["price"]
```

```python
# 2. Apply Recursive Feature Elimination (RFE)
model = LinearRegression()
selector = RFE(estimator=model, n_features_to_select=3)

selector = selector.fit(X, y)
```

```python
# 3. Get selected feature names
selected_features = X.columns[selector.get_support()]

print("Top Features Selected by RFE:")
print(selected_features)
```

```python
# Optional: Print ranking of all features (1 = selected)
print("\nFeature Rankings:")
for feature, rank in zip(X.columns, selector.ranking_):
    print(f"{feature:25} Rank: {rank}")
```
