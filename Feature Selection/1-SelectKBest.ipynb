# SelectKBest on a Custom Housing Dataset

```python
# Import the libraries
import numpy as np
import pandas as pd
from sklearn.feature_selection import SelectKBest, f_regression
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

### Create a dataset

```python
# Create a dataset
np.random.seed(42)
num_samples = 500

data = {
    "house_size": np.random.normal(2000, 500, num_samples),             # sqft
    "num_rooms": np.random.randint(3, 8, num_samples),                  # rooms
    "crime_rate": np.random.uniform(0, 1, num_samples),                 # 0 = safe, 1 = dangerous
    "distance_to_city_center": np.random.uniform(1, 30, num_samples),   # miles
    "school_rating": np.random.uniform(1, 10, num_samples),             # 1-10
    "age_of_house": np.random.randint(1, 80, num_samples),              # years
}

df = pd.DataFrame(data)
df.head()
```

```python
# Create a target variable: House Price
df["price"] = (
    df["house_size"] * 150
    + df["school_rating"] * 4000
    - df["crime_rate"] * 30000
    - df["age_of_house"] * 500
    - df["distance_to_city_center"] * 1500
    + np.random.normal(0, 20000, num_samples)   # noise
)
```

```python
df.head()
```

### Split the data

```python
X = df.drop("price", axis=1)
y = df["price"]
```

### Apply SelectKBest

```python
# Create a SelectKBest object.
# Use f-regression as the statistical test
# k=3 features, after scoring all features, keep only the 3 highest scoring ones. So selector is now a feature-selection model
selector = SelectKBest(score_func=f_regression, k=3)
selector
```

```python
# It calculates the F-score for each variable in X, ranks all the features and then decides which 3 features to keep
# .transform(X) --> Creates a new dataset containing only those selected 3 features and removes all other non-selected features
X_new = selector.fit_transform(X, y)
```

```python
# After I run fit or fit_transform, the selector knows which features were chosen. .get_support() returns a boolean mask indicating which features were selected
selected_features = X.columns[selector.get_support()]
selected_features
```

```python
print(f"Top Selected Features: {selected_features}")
```
