## loc

```python
student_dict = {
    'name':['nitish','ankit','rupesh','rishabh','amit','ankita'],
    'iq':[100,90,120,80,0,0],
    'marks':[80,70,100,50,0,0],
    'package':[10,7,14,2,0,0]
}

students = pd.DataFrame(student_dict)
students.set_index('name',inplace=True)
students
```

```python
# Fetching just one row
students.loc["nitish"]
```

```python
# Fetching multiple rows
students.loc[["nitish", "ankit"]]
```

```python
students.loc["rupesh", "marks"]
```

# Filtering a Dataframe

```python
# Find all the final winners
df1.head(2)
```

```python
df1.loc[(df1["MatchNumber"] == "Final"), ["Season", "WinningTeam"]]
```

```python
# How many super over finishes have occurred?
df1[df1["SuperOver"]=="Y"].shape[0]
```

```python
df1["WinningTeam"].unique()
```

```python
# How many matches CSK has won in Kolkata
df1[(df1["WinningTeam"] == "Chennai Super Kings") & (df1["City"] == "Kolkata")].shape[0]
```

```python
# Toss winner is match winner in percentage
round(((df1[df1["TossWinner"] == df1["WinningTeam"]].shape[0])/len(df1)) * 100)
```
