# Missing data Handling

```python
students = pd.DataFrame(
    {
        'name':['nitish','ankit','rupesh',np.nan,'mrityunjay',np.nan,'rishabh',np.nan,'aditya',np.nan],
        'college':['bit','iit','vit',np.nan,np.nan,'vlsi','ssit',np.nan,np.nan,'git'],
        'branch':['eee','it','cse',np.nan,'me','ce','civ','cse','bio',np.nan],
        'cgpa':[6.66,8.25,6.41,np.nan,5.6,9.0,7.4,10,7.4,np.nan],
        'package':[4,5,6,np.nan,6,7,8,9,np.nan,np.nan]})
students
```

## isnull

```python
students.isnull()
```

## notnull

```python
students.notnull()
```

## dropna(series + dataframe) - Remove missing values

```python
# dropna on Series
students["name"].dropna()
```

```python
# dropna on Dataframe
students.dropna()
```

## fillna (series + dataframe) --> Fill Na/NaN values with value

```python
# # Replace a single value ---->  df.replace(0, np.nan)                                  Replace all 0's with NaN
# Replace multiple values with a single value ---->  df.replace([0, -2], np.nan)       Replace both 0 and -2 with NaN
# Replace Strings  ---->   df.replace({"HR": "Useless department ever to exist"})
# Replace with a dictionary ---->  df.replace({'Age': {25: 26}, 'Salary': {60000: 62000}})    Change age 25 to 26, And Salary 60000 to 62000
```

```python
students.head()
```

```python
# fillna on Series - Fill missigng values of cgpa with the avg cgpa
students["cgpa"] = students["cgpa"].fillna(students["cgpa"].mean())
students
```

```python
# fillna on Dataframe - Replace all missing values with 0 (does not makes sense)
students.fillna(0)
```

## drop_duplicates (series + dataframe) --> Returns dataframe with duplicated rows removed

```python
df = pd.DataFrame({
        "brand": ["Yum Yum", "Yum Yum", "Indomie", "Indomie", "Indomie"],
        "style": ["cup", "cup", "cup", "pack", "pack"],
        "rating": [4, 4, 3.5, 15, 5]})
df
```

```python
# By default, it removes duplicate rows based on all columns
df.drop_duplicates()

# To remove duplicates based on specific column(s), use subset
# To remove duplicates and keep last occurrence, use keep
```
