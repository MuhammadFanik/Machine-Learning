# Editing Series

```python
marks = pd.Series([75, 74, 63, 51, 101], index=["English", "Maths", "Science", "History", "Debate on Babar Azam"])
marks
```

```python
# Using Indexing - Change English marks to 85
marks["English"] = 85
marks
```

# Series with Python Functionalities

```python
# len/type/dir/sorted/max/min
print(f"Length is: {len(marks)}")
print(f"Type is: {type(marks)}")
print(f"Sorted: {sorted(marks)}")
print(f"Max: {max(marks)}")
print(f"Min: {min(marks)}")
```

```python
# Type conversion
list(marks)
```

```python
# membership operator - It works on the index, not the values
print("English" in marks)  # This only works on the index
print(101 in marks.values)  # This will work on the values
```

```python
# Looping

# This will loop over the values
for i in marks:
    print(i)
```

```python
# This will loop over the index
for i in marks.index:
    print(i)
```
