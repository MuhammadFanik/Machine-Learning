# Comparisons

```python
# < (less than)
print(x<5)
```

```python
# > (Greater than)
print(x>4)
```

```python
# <= (less than equal to)
print(x<=5)
```

```python
# >= (greater than equal to)
print(x>=5)
```

```python
# != (not equal)
print(x!=2)
```

```python
# == (equal)
print(x==4)
```

## Working with Boolean Arrays

```python
x = np.array([[5, 0, 3, 3],
              [7, 9, 3, 5],
              [2, 4, 7, 6]])
```

```python
# Count the number of True entries
print(np.count_nonzero(x<6))
print(np.sum(x<6, axis=1))
```

```python
print(np.any(x>8))  # Any value > 8
print(np.any(x<0))  # Any value < 0
print(np.all(x<10)) # Are all values < 10
print(np.all(x==10))    # Are all values equal to 10
```
