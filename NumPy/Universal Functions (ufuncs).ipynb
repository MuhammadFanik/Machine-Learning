# Array Arithmetic

```python
x = np.arange(4)
print(f"x: {x}")
print(f"x-5: {x-5}")       # Subtraction
print(f"x+5: {x+5}")       # Addition
print(f"x*2: {x*2}")       # Multiplication
print(f"x/2: {x/2}")       # Division
print(f"x//2: {x//2}")     # Floor Division
print(f"-x: {-x}")         # Negation
print(f"x**3: {x**3}")     # Exponent
print(f"x%2: {x%2}")       # Modulus
```

```python
# Absolute Value
x = np.array([-3, 4, 7, -2, -11])
print(np.absolute(x))
```

# Trigonometric Functions

```python
theta = np.linspace(0, np.pi, 3)
print(theta)
print(f"Sin Theta: {np.sin(theta)}")        # SIn theta
print(f"Cos Theta: {np.cos(theta)}")        # Cos Theta
print(f"Tan Theta: {np.tan(theta)}")        # Tan theta

# Inverse Trig functions --> arcsin, arccos, arctan
```

## Exponents and Logarithms

```python
x = [1, 2, 3]
print(f"x: {x}")
print(f"e^x: {np.exp(x)}")
print(f"2^x: {np.exp2(x)}")
print(f"3^x: {np.power(x, 3)}")
print("ln(x) =", np.log(x))
print("log2(x) =", np.log2(x))
print("log10(x) =", np.log10(x))
print("exp(x) - 1 =", np.expm1(x))
print("log(1 + x) =", np.log1p(x))
```
