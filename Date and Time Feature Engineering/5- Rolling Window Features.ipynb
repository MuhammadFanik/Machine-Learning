```python
import numpy as np
import pandas as pd
import datetime as dt
```

```python
df = pd.DataFrame({
    'date': pd.date_range(start='2024-01-01', periods=10, freq='D'),
    'sales': [200, 220, 180, 240, 210, 260, 230, 250, 270, 290]})

df
```

```python
# Sales_rolling _mean_3
df["sales_rolling_mean_3"] = df["sales"].rolling(window=3).mean()
df
```

```python
# sales_rolling_mean_5
df["sales_rolling_mean_5"] = df["sales"].rolling(window=5).mean()
df
```

```python
# sales_rolling_sum_3
df["sales_rolling_sum_3"] = df["sales"].rolling(window=3).sum()
df
```
