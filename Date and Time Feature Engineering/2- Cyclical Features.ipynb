```python
import numpy as np
import pandas as pd
import datetime as dt
```

```python
df = pd.DataFrame({
    'event_id': [1, 2, 3, 4, 5, 6],
    'timestamp': [
        '2024-01-01 08:00:00',
        '2024-03-15 14:00:00',
        '2024-06-21 20:00:00',
        '2024-09-10 03:00:00',
        '2024-11-30 23:00:00',
        '2024-12-31 12:00:00']
})
df["timestamp"] = pd.to_datetime(df["timestamp"])
df
```

```python
# Extract the month and hour
df["month"] = df["timestamp"].dt.month
df["hour"] = df["timestamp"].dt.hour

df
```

```python
# Apply Cyclical Encoding

# Monthly Cyclical Encoding
df["month_sin"] = np.round(np.sin(2*np.pi * df["month"] / 12), 3)
df["month_cos"] = np.round(np.cos(2*np.pi * df["month"] / 12), 3)

# Hourly Cyclical Encoding
df["hour_sin"] = np.round(np.sin(2*np.pi * df["hour"] / 24), 3)
df["hour_cos"] = np.round(np.cos(2*np.pi * df["hour"] / 24), 3)

df
```
