```python
import numpy as np
import pandas as pd
import datetime as dt
```

```python
today = pd.Timestamp('2024-12-31')
```

```python
df = pd.DataFrame({
    'customer_id': [1, 2, 3, 4, 5],
    'last_purchase_date': [
        '2024-12-01',
        '2024-09-15',
        '2024-01-10',
        '2024-11-20',
        '2023-06-30'
    ],
    'contract_expiry_date': [
        '2025-01-15',
        '2025-06-30',
        '2024-12-15',
        '2025-03-01',
        '2025-02-28']})

df['last_purchase_date'] = pd.to_datetime(df["last_purchase_date"])
df["contract_expiry_date"] = pd.to_datetime(df["contract_expiry_date"])

df
```

### Create Time Since Features

```python
# days since last purchase
df['days_since_last_purchase'] = (today - df['last_purchase_date']).dt.days
df
```

### Create Time Until Features

```python
# days until contract expiry
df["days_until_contract_expiry"] = (df["contract_expiry_date"] - today).dt.days
df
```
