```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
```

```python
# Seed
np.random.seed(0)

# Generate random data
X = np.linspace(0, 10, 100).reshape(-1, 1)  # (100, 1) --> Shape. Converts it into a column vector
y = 2 + 0.5 * X**2 - 3*X + np.random.randn(100, 1) * 3  # Quadratic with noise
```

```python
plt.scatter(X, y)
plt.grid()
plt.show()
```

```python
# Fit Linear Regression
lin_reg = LinearRegression()  # Creating a linear model
lin_reg.fit(X, y)   # Trains the model
```

```python
y_pred_linear = lin_reg.predict(X)
```

```python
# Fit Polynomial Regression with degree = 2
poly = PolynomialFeatures(degree=2) # Create polynomial terms
X_poly = poly.fit_transform(X)  # Applies the transformation to all data

poly_reg = LinearRegression()
poly_reg.fit(X_poly, y)

y_pred_poly = poly_reg.predict(X_poly)
```

```python
# Graphs

plt.figure(figsize=(12, 5))

# Linear Regression plot
plt.subplot(1, 2, 1)
plt.scatter(X, y, color="blue", label="Data")
plt.plot(X, y_pred_linear, color="red", label="Linear Fit")
plt.title("Linear Regression")
plt.legend()

# Polynomial Regression plot
plt.subplot(1, 2, 2)
plt.scatter(X, y, color="blue", label="Data")
plt.plot(X, y_pred_poly, color="green", label="Polynomial Fit (deg=2)")
plt.title("Polynomial Regression")
plt.legend()

plt.show()
```
