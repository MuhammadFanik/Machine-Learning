```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Import the dataset
df = pd.read_csv("placement.csv")
df.head()
```

```python
# Plot the scatter plot --> Data is somewhat linearly related
plt.scatter(df["cgpa"], df["package"])
plt.grid()
plt.show()
```

```python
# Separating the input and output Columns
X = df[["cgpa"]]
y = df["package"]
```

```python
# Train Test Split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
```

```python
# Train a Simple linear Regression Model
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(X_train, y_train)
```

```python
# Extract the parameters
print(lr.coef_[0])     # gradient
print(lr.intercept_)    # y_intercept
```

```python
# Make predictions
predicted_salary = lr.predict([[8.5]])  # Scikit learn expects the input to .predict  to be a 2d array
print(predicted_salary[0])
```

```python
# Plot the regression line
plt.scatter(X, y, color="blue")
plt.plot(X, lr.predict(X), color="red") # Regression line
plt.xlabel("CGPA")
plt.ylabel("Package")
plt.title("Regression line on CGPA vs Package")
plt.grid()
plt.show()
```

# My Task

```python
# Years of Experience (Input), performance Score (output)
# Your job is to build a model to predict performance score based on experience.

df = pd.read_csv("experience_performance.csv")
df.head()
```

```python
# Visualize the data
plt.scatter(df["experience"], df["performance_score"])
plt.grid()
plt.show()
```

```python
# Separate the columns
X = df[["experience"]]
y = df["performance_score"]
```

```python
# Train Test the data
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.75, random_state=0)
```

```python
lr = LinearRegression()
lr.fit(X_train, y_train)
```

```python
# Output the parameters
m = lr.coef_[0]
b = lr.intercept_
print(m)
print(b)
```

```python
# Make predictions
pred_value1 = lr.predict([[3.5]])[0]
pred_value2 = lr.predict([[7.0]])[0]
print(f"Predicted performance score for 3.5 years is {pred_value1:.2f}")
print(f"Predicted performance score for 7 years is {pred_value2:.2f}")
```

```python
# Plot the regression line
plt.scatter(X, y, color="blue")
plt.plot(X, lr.predict(X), color="orange")
plt.xlabel("Experience in years")
plt.ylabel("Performance Score")
plt.title(f"Simple Linear Regression Equation y = {m:.2f}x + {b:.2f}")
plt.grid()
plt.show()
```
