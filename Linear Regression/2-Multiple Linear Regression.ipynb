```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
# Import the dataset
df = pd.read_csv("multiple_regression_data.csv")
df.head()
```

```python
df.shape
```

```python
# Separate the input and output cols
X = df[["study_hours", "experience_years", "projects_completed"]]
y = df["performance_score"]
```

```python
# Train test split
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.7, random_state=42)
```

```python
# Create and train MLR
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
# Step 5: Display model coefficients
print("Intercept (β₀):", model.intercept_)
print("Coefficients (β):", model.coef_)
print("Feature names:", X.columns.tolist())
```

## Make Predictions

```python
# Make predictions on my custom Values set
prediction = model.predict([[5, 3, 7]])
print(prediction[0])
```

```python
# Prediction on the test set
prediction1 = model.predict(X_test)
print(prediction1)
```

```python
from sklearn.metrics import mean_squared_error
mse = mean_squared_error(y_test, prediction1)
print(f"Mean Squared Error: {mse:.2f}")
```
