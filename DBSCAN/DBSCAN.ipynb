```python
from sklearn.cluster import DBSCAN
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([[1, 2], [2, 2], [2, 3],[8, 7], [8, 8], [25, 80]])
X
```

```python
plt.scatter(X[:, 0], X[:, 1])
plt.show()
```

```python
# Creates a DBSCAN object
# eps=3 --> The maximum distance between two points for them to be considered neighbors
# min_samples = 2 --> The min number of points required to form a dense region
db = DBSCAN(eps=3, min_samples=2)

# Trains the DBSCAN algorithm on the dataset X. The algorithm identifies cluster based on density
db.fit(X)

# Accesses the cluster labels assigned to each point in X
# Returns an array where each element corresponds to a data point's cluster assignment. -1= Noise
db.labels_
```
